```python

from pathlib import Path
import joblib
import pandas as pd

ROOT = Path.cwd()

# Your notebook is currently inside the models folder.
if ROOT.name.lower() != "models":
    ROOT = ROOT / "models"

MODEL_PATH = ROOT / "jio_churn_model.pkl"

if not MODEL_PATH.exists():
    raise FileNotFoundError(f"Jio model not found: {MODEL_PATH}")

model = joblib.load(MODEL_PATH)

print("Jio model loaded successfully!")
print("Model type:", type(model))
print("Model path:", MODEL_PATH)
```

```python

features = getattr(model, "feature_names_in_", None)

if features is None and hasattr(model, "named_steps"):
    for step in model.named_steps.values():
        if hasattr(step, "feature_names_in_"):
            features = step.feature_names_in_
            break

if features is None:
    print("Could not automatically find the input headings.")
    print("Pipeline steps:", getattr(model, "named_steps", None))
else:
    features = list(features)
    print("Required customer input fields:")
    for i, feature in enumerate(features, start=1):
        print(f"{i}. {feature}")
```
